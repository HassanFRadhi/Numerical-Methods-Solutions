```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
```

```python
c = '0.5 0.8 1.5 2.5 4'
k = '1.1  2.4  5.3  7.6  8.9'
c = np.array([float(i) for i in c.split()])
k = np.array([float(i) for i in k.split()])
```

```python
plt.scatter(1/(c**2), 1/k)
plt.xlim(0)
plt.ylim(0)
```

```python
def model(c,kmax,cs):
    return (kmax*c**2)/(c**2+cs)
popt, _ = curve_fit(model,c,k,p0=[10,2.5])
k2 = (popt[0] * c**2)/(c**2+popt[1])
plt.scatter(c,k)
plt.plot(c,k2,'r')
# At c = 2
(popt[0] * 2**2)/(2**2+popt[1]) 
```
