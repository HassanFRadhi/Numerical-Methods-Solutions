```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
```

```python
x = '0.4  0.8  1.2  1.6  2  2.3'
y = '800  975  1500  1950  2900  3600'
x = np.array([float(i) for i in x.split()])
y = np.array([float(i) for i in y.split()])
```

```python
plt.scatter(x,np.log(y))
plt.ylim(0)
plt.xlim(0)
np.exp(6)
```

```python
def model(x,a,b):
    return a * np.exp(b*x)

popt, _ = curve_fit(model,x,y,p0=[403,0.8])
y2 = popt[0] * np.exp(popt[1] * x)
plt.scatter(x,y)
plt.plot(x,y2,'r')
popt[1]
```

```python
plt.scatter(x,np.log(y))
plt.plot(x,np.log(y2),'r')
plt.title(r'$y = a*exp(bx)$', pad=12)
plt.ylabel('ln [y]', size=20)
plt.xlabel('x', size=20)
plt.grid()
```
