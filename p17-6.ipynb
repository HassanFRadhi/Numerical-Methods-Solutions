```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.optimize as si
```

```python
xi = '''1  2  3  4  5  6  7  8  9'''
yi = '''1 1.5  2  3  4  5  8  10  13'''
x = np.array([int(i) for i in xi.split()])
y = np.log(np.array([float(i) for i in yi.split()]))
yd = np.array([float(i) for i in yi.split()])
plt.scatter(x,y)
np.exp(-0.3)
B = (y[1] - y[0]) / (x[1] - x[0])
```

```python
def model(x,a,b):
    return a * np.exp(b* x)
popt, _ = si.curve_fit(model,x,yd,p0=[0.7,0.4]) #first guess p0 is constant
ymodel = popt[0] * np.exp(popt[1] * x)
plt.plot(x,ymodel,'r')
plt.scatter(x,yd)
```

```python
r2e = (yd-ymodel)**2
seoee = np.sqrt(sum(r2e)/(len(x) - 2))
seoee
```

```python
def linearmodel(x,a,b):
    return a + b * x
popt2, _ = si.curve_fit(linearmodel,x,yd,p0=[0.7,0.4])
ylmodel = popt2[0] + popt2[1] * x
plt.scatter(x,yd)
plt.plot(x,ylmodel,'r')
```

```python
r2l = (yd-ylmodel)**2
seoel = np.sqrt(sum(r2l)/(len(x) - 2))
seoel
```

```python
def polymodel(x,a,b,c):
    return a + b * x + c * x**2
popt3, _ = si.curve_fit(polymodel,x,yd,p0=[0.7,0.4,1])
ypmodel = popt3[0] + popt3[1] * x + popt3[2] * x**2
plt.scatter(x,yd)
plt.plot(x,ypmodel,'r')
```

```python
r2p = (yd-ypmodel)**2
seoep = np.sqrt(sum(r2p)/(len(x) - 2))
seoep
```
