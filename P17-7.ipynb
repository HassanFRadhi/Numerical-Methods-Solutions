```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.optimize as si
```

```python
x = '0.75  2  3  4  6  8  8.5'
y = '1.2  1.95  2  2.4  2.4  2.7  2.6'
x = np.array([float(i) for i in x.split()])
y = np.array([float(i) for i in y.split()])
```

```python
def model1(x,a,b):
    return (a*x)/(b+x)

def model2(x,a,b):
    return a * x**b

def model3(x,a,b,c):
    return a * x**2 + b * x + c

```

```python
plt.plot(x,y)
plt.xlim(0)
plt.ylim(0)
```

```python
popt1, _ = si.curve_fit(model1,x,y,p0=[0.75,0.4])
popt2, _ = si.curve_fit(model2,x,y,p0=[0.75,0.4])
popt3, _ = si.curve_fit(model3,x,y,p0=[0.75,0.4,0.75])
```

```python
y1 = (popt1[0]*x)/(popt1[1]+x)
y2 = popt2[0] * x**popt2[1]
y3 = popt3[0] * x**2 + popt3[1] * x + popt3[2]
```

```python
fig,axes = plt.subplots(3)
axes[0].plot(x,y1,'r')
axes[0].set_title(r'$ax/(b+X)$')
axes[1].plot(x,y2,'k')
axes[1].set_title(r'$ax^b$')
axes[2].plot(x,y3,'b')
axes[2].set_title(r'$ax^2 + bx + c$')
```
