```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
```

```python
x = '0.4  0.8  1.2  1.6  2  2.3'
y = '800  975  1500  1950  2900  3600'
x = np.array([float(i) for i in x.split()])
y = np.array([float(i) for i in y.split()])
```

```python
def model(x,a,b):
    return a * 10**(b*x)
popt, _ = curve_fit(model,x,y,p0=[403,0.8])
popt
y2 = popt[0] * 10**(popt[1] * x)
plt.scatter(x,y)
plt.plot(x,y2,'r')
popt[1]
```
