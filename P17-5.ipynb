```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.optimize as si
```

```python
xi = '''2  4  6  7  10  11  14  17  20'''
yi = '''1  2  5  2  8  7  6  9  12'''
x = np.array([int(x) for x in xi.split()])
y = np.array([(int(y) - np.e) for y in yi.split()])
plt.scatter(x,y)
plt.xlim(0)
```

```python
def model(x,a):
    return a*x
popt, _ = si.curve_fit(model,x,y,p0=[1.1])
ymodel = popt[0] * x
plt.plot(x,ymodel,'r')
plt.xlim(0)
plt.ylim = (0)
plt.scatter(x,y)
popt
np.roots(x,ymodel)
```
