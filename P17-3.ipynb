```python
import numpy as np
import scipy.optimize as si
import matplotlib.pyplot as plt
```

```python
xi = '''0  2  4  6  9  11  12  15  17  19 '''
x = np.array([int(x) for x in xi.split()])
yi = '''5  6  7  6  9  8  7  10  12  12 '''
y = np.array([int(y) for y in yi.split()])
```

```python
def fy(y,a,b):
    return a + b*y
popt, popv = si.curve_fit(fy,y,x,p0=[-1,2.5])

xpredict = popt[0] + popt[1] * np.linspace(0,13,100)

popt
```

```python
plt.scatter(y,x)
plt.plot(np.linspace(0,13,100),xpredict)
plt.grid()
```
