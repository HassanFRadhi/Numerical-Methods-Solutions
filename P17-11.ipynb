```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
```

```python
x = '0.1  0.2  0.4  0.6  0.9  1.3  1.5  1.7  1.8'
y = '0.75  1.25  1.45  1.25  0.85  0.55  0.35  0.28  0.18'
x = np.array([float(i) for i in x.split()])
y = np.array([float(i) for i in y.split()])
```

```python

#plt.scatter(x,y,s=5)
plt.plot(x,np.log(y/x),'r')
plt.ylim(0)
plt.xlim(-1)
np.exp(2)
```

```python
def model(x,a,b):
    return a*x*np.exp(x*b)

popt, _ = curve_fit(model,x,y,p0=[7,-3])
y2 = popt[0] * x * np.exp(x * popt[1])
plt.scatter(x,y)
plt.plot(x,y2,'r')
```
